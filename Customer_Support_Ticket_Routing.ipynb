```python
import spacy

# STEP 1: NLP PIPELINE

nlp = spacy.load("en_core_web_sm")

tickets = [
    "My payment failed.",
    "Unable to login to my account.",
    "Need refund for my order.",
    "I want to buy a new product."
]


# STEP 2: SUPPORT CATEGORIES

categories = {
    "Billing Team": [
        "payment",
        "billing",
        "charged",
        "invoice",
        "card",
        "Due Date",
        "Late Payment charges",
        "failed payment"
    ],

    "Technical Team": [
        "login",
        "password",
        "error",
        "Network Issue",
        "technical",
        "account",
        "unable to login"
    ],

    "Refund Team": [
        "refund",
        "return",
        "money back",
        "cancel order"
    ],

    "Sales Team": [
        "buy",
        "purchase",
        "product",
        "price",
        "sales"
    ]
}


# STEP 3: PROCESS EACH TICKET

for ticket in tickets:

    print("\n-----------------------------")
    print("Ticket:", ticket)

    doc = nlp(ticket)

    # Convert ticket to lowercase
    ticket_text = ticket.lower()

    scores = {}


    # STEP 4: KEYWORD MATCHING

    for category, keywords in categories.items():

        matched_words = []

        for keyword in keywords:

            if keyword.lower() in ticket_text:
                matched_words.append(keyword)

        scores[category] = len(matched_words)


    # STEP 5: ENTITY RECOGNITION

    entities = []

    for ent in doc.ents:
        entities.append((ent.text, ent.label_))


    # STEP 6: FIND BEST CATEGORY

    best_category = max(scores, key=scores.get)

    # If no keyword is found
    if scores[best_category] == 0:
        best_category = "Sales Team"


    print("Scores:", scores)
    print("Entities:", entities)
    print("Routed To:", best_category)

```

```python
# Interactive mode: Get custom ticket input from the user
user_ticket = input("Enter your support ticket description: ")

if user_ticket.strip():
    print("\n--- Processing Your Ticket ---")
    print("Ticket:", user_ticket)

    # Process with SpaCy
    doc = nlp(user_ticket)
    ticket_text = user_ticket.lower()
    scores = {}

    # Keyword Matching
    for category, keywords in categories.items():
        matched_words = []
        for keyword in keywords:
            if keyword.lower() in ticket_text:
                matched_words.append(keyword)
        scores[category] = len(matched_words)

    # Entity Recognition
    entities = [(ent.text, ent.label_) for ent in doc.ents]

    # Find Best Category
    best_category = max(scores, key=scores.get)
    if scores[best_category] == 0:
        best_category = "Sales Team"

    print("Scores:", scores)
    print("Entities:", entities)
    print("Routed To:", best_category)
else:
    print("No ticket entered.")
```
